```python
# 1) Average ticket price for each fare condition 
```

```python
query = """
SELECT fare_conditions,
       ROUND(AVG(amount), 2) AS avg_price
FROM ticket_flights
GROUP BY fare_conditions
ORDER BY fare_conditions;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT departure_airport,
       SUM(CASE WHEN status = 'Delayed' THEN 1 ELSE 0 END) AS n_delays
FROM flights
GROUP BY departure_airport
ORDER BY n_delays DESC LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
# 3) The least popular destinations (airports with the lowest number of arrivals)
```

```python
query = """
SELECT arrival_airport,
       COUNT(flight_id) AS n_flights
FROM flights
GROUP BY arrival_airport
ORDER BY n_flights ASC LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT departure_airport,
       COUNT(flight_id) AS n_flights
FROM flights
GROUP BY departure_airport
ORDER BY n_flights DESC LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
big_data = """
SELECT *
FROM bookings
JOIN tickets
    ON bookings.book_ref = tickets.book_ref
JOIN ticket_flights
    ON tickets.ticket_no = ticket_flights.ticket_no
JOIN flights
    ON ticket_flights.flight_id = flights.flight_id LIMIT 10;
"""

pd.read_sql(big_data, conn)
```

```python
# 5) Aircraft model which the highest number of flights 
```

```python
query = """
SELECT flights.aircraft_code,
       aircrafts_data.model AS aircraft_model,
       COUNT(flights.flight_id) AS n_flights
FROM flights
JOIN aircrafts_data
    ON flights.aircraft_code = aircrafts_data.aircraft_code
GROUP BY flights.aircraft_code, aircrafts_data.model
ORDER BY n_flights ASC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
# The results show that the CN1 and CR2 aircraft carry the most flights, making them the most popular
```

```python
# 6) Finding the  top 10 most loyal customers
```

```python
query = """
SELECT passenger_id,
       COUNT(ticket_no) AS n_tickets
FROM tickets
GROUP BY passenger_id
ORDER BY n_tickets DESC LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
# The result suggests either we have no loyal customers (unlikely),
# or the data covers a short period where even loyal customers took only one flight
```

```python
# 7) aircrafts generating the highest revenue
```

```python
query = """
SELECT flights.aircraft_code,
       aircrafts_data.model AS aircraft_model,
       SUM(ticket_flights.amount) AS total_revenue
FROM flights
JOIN ticket_flights
    ON flights.flight_id = ticket_flights.flight_id
JOIN aircrafts_data
    ON flights.aircraft_code = aircrafts_data.aircraft_code
GROUP BY flights.aircraft_code, aircrafts_data.model
ORDER BY total_revenue DESC;
"""

pd.read_sql(query, conn)
```

```python
# As we can see, the SU9 is the highest revenue-generating aircraft for our customers.
```

```python
# Finally, I would like to determine the proportion of flights that were business class for each aircraft.
```

```python
query = """
SELECT flights.aircraft_code,
       aircrafts_data.model AS aircraft_model,
       SUM(CASE WHEN ticket_flights.fare_conditions = 'Business' THEN 1 ELSE 0 END) * 1.0 / COUNT(ticket_flights.ticket_no) AS business_rate
FROM flights
JOIN ticket_flights
    ON flights.flight_id = ticket_flights.flight_id
JOIN aircrafts_data
    ON flights.aircraft_code = aircrafts_data.aircraft_code
GROUP BY flights.aircraft_code, aircrafts_data.model
ORDER BY business_rate DESC;
"""

pd.read_sql(query, conn)
```

```python
# The chart is very interesting. My hypothesis was that SU9 leads in revenue generation
# because most of its tickets are business class. However, there are three aircraft with a higher proportion of business class passengers,
# yet these aircraft lag behind in revenue compared to the top three.
```
