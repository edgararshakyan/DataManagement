```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

df = []

for page in range(1,11):
    url = "https://quotes.toscrape.com/"
    page_url = url + "page/" + str(page)
    html = requests.get(page_url).text
    soup = BeautifulSoup(html, 'html.parser')
    quote_divs = soup.find_all("div", class_="quote")


    for quote in quote_divs:
        
        quote_text = quote.span.string
        qoute_author = quote.small.string
        
        qoute_tags = [t.get_text() for t in quote.find_all("a", class_="tag")]
        
        auther_url = quote.find("a",)['href']
        about_url = url + auther_url
        
        author_html = requests.get(about_url).text
        author_soup = BeautifulSoup(author_html, "html.parser")

        auther_born = author_soup.find("span","author-born-date").text
        auther_description = author_soup.find("div", "author-description").text
                
        df.append({
            'quote': str(quote_text),
            'author': str(qoute_author),
            'tags': list(qoute_tags),
            'born': auther_born,
            'description': auther_description
        })
    print(page)
    print(page_url)
    
df = pd.DataFrame(df)
df.to_csv("quotes.csv", index=False)
print(df)
```
